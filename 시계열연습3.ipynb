```python
# 전체 시계열을 앞 절반과 뒤 절반으로 나눕니다.
# 앞 절반의 이동 표준편차 평균과 뒤 절반의 이동 표준편차 평균을 비교합니다.
# 뒤 절반 평균이 더 크면 → '증가'
# 아니면 → '감소'
```

```python
# ── [실습 6] 여러 센서 추세·변동성 자동 판정 ────────────────────────────────
for col in ['s_11', 's_12', 's_3']:
    ma = e2[col].rolling(5).mean()
    tr = '상승' if ma.dropna().iloc[-1] > ma.dropna().iloc[0] else '하강'
    s = e2[col].rolling(5).std(); h = len(e2) // 2
    vol = '증가' if s.iloc[h:].mean() > s.iloc[:h].mean() else '감소'
    print(col, '추세', tr, '변동성', vol)             # s_11 상승·증가 / s_12 하강·감소 / s_3 상승·증가
```

```python
# ── [실습 7] 이상 의심 구간과 정비 리포트 — 캡스톤 ──────────────────────────
g = e2.copy(); g['std5'] = g['s_11'].rolling(5).std()
thr = g['std5'].mean() + g['std5'].std()
sus = g[g['std5'] > thr]
print('의심 cycle 개수:', len(sus),
      '/ 범위:', int(sus['time_cycles'].min()), '~', int(sus['time_cycles'].max()))   # 38 / 11 ~ 283
print(f'[리포트] 엔진 2, {len(e2)} cycle, s_11 상승·s_7 반대, 의심 {len(sus)}곳 → 우선 점검')
```
